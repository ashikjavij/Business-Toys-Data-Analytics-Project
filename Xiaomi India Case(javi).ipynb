```python
#Step to build a predictive Model
#1.Creating a variable for X & Y
#2.Import and build linear regression Model
#3.Fitting the Model
#4.Predicting the new y variables
#5.Evaluating the Model
```

```python
#Demontrate a the 5 step used to build a predictive Model using an examle
```

```python
#1.Creating a variable for X & Y
df.head()
```

```python
#Creating the X variable 

x= df.iloc[:,3:6]
x.head()
```

```python
#Creating the y variable
y= df['Amount Spent (Rs.)']
y.head()
```

```python
y=df.iloc[:,6:7]
y.head()
```

```python
y=df.iloc[:,6:7]
y.head()
```

```python
#2.Import and build linear regression Model
lr =LinearRegression()
```

```python
lr.fit(x,y)
```

```python
#4.Predicting the new y variables
y_pred = lr.predict(x)
y_pred
```

```python
#5.Evaluating the Model using r square (d)
r2= r2_score(y,y_pred)
r2
```

```python
#Printing the final result
print("The R Square value for the model is", round(r2,2)*100,"%")
```

```python
#Note:
# R-squared (r2) is used to evaluate the goodness of fit for a statistical model 
# It represents the proportion of the variance in the dependent variable thet's explained by the independent variables in the model
#A higher r2 indicate that the models prediction are closer to the actual data points
```
